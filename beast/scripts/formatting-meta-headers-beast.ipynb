```python
df['states_grouped'] = np.select(
    [df['division'] == 'Pennsylvania', df['division'] == 'New York', df['division'] == 'New Jersey'],
    ['PA', 'NY', 'NJ'],
    default=df['geo_region_na']
)
df
```

```python
df['wild_dom_lbm'].unique()
```

```python
df['wild_dom_lbm'] = df['wild_dom_lbm'].fillna("unknown")
df['wild_dom_lbm'].unique()
```

```python
##new column called sate-domstat
##if domstat = unknown, then state-domstat = "other
##if domstat = wild, then state-domstat = "wild
##if domstat = lbm OR domestic and states_grouped = PA, then state_domstat = "PA-domestic"
##if domstat = domestic and states-grouped = NY, then state-domstat = NY-domestic"

df["state_domstat"] = np.select(
    [
        df["wild_dom_lbm"].isna() | (df["wild_dom_lbm"] == "unknown"), ##is unknown or NA domstat
        (df["wild_dom_lbm"] == "domestic") & (~df["states_grouped"].isin(["NY", "PA"])), ##not in PA or NY and is domestic
        df["wild_dom_lbm"] == "wild", ##is wild
        (df["wild_dom_lbm"].isin(["lbm", "domestic"])) & (df["states_grouped"] == "PA"), ##lbm or domestic and from PA
        (df["wild_dom_lbm"] == "domestic") & (df["states_grouped"] == "NY") ##domestic NY
    ],
    [
        "other",
        "other",
        "wild",
        "PA_domestic",
        "NY_domestic"
    ],
    default=None
)
```

```python
df['state_domstat'].value_counts()
```

```python
df["state_domstat2"] = np.select(
    [
        (df["wild_dom_lbm"] == "unknown") & (~df["states_grouped"].isin(["NY", "PA"])), ##is unknown or NA domstat and not PA or NY
        (df["wild_dom_lbm"] == "domestic") & (~df["states_grouped"].isin(["NY", "PA"])), ##not in PA or NY and is domestic
        df["wild_dom_lbm"] == "wild", ##is wild
        (df["wild_dom_lbm"].isin(["lbm", "domestic"])) & (df["states_grouped"] == "PA"), ##lbm or domestic and from PA
        (df["wild_dom_lbm"] == "domestic") & (df["states_grouped"] == "NY"), ##domestic NY
        (df["wild_dom_lbm"] == "unknown") & (df["states_grouped"] == "NY"), ##unknown NY
        (df["wild_dom_lbm"] == "unknown") & (df["states_grouped"] == "PA") ##unknown NY

    ],
    [
        "other",
        "other",
        "wild",
        "PA_domestic",
        "NY_domestic",
        "NY_unknown",
        "PA_unknown"
    ],
    default=None
)
df['state_domstat2'].value_counts()
```

```python
df["pa_ny_nj_other"] = np.select(
    [
        (~df["states_grouped"].isin(["NY", "PA","NJ"])), ##is unknown or NA domstat and not PA or NY
    ],
    [
        "other"
    ],
    default=df['states_grouped']
)
df['pa_ny_nj_other'].value_counts()
```

```python
# Create a dictionary for quick lookup of metadata by subtree_name
metadata_dict = df.set_index('strain').to_dict(orient='index')

# Process all subtree alignment files in a folder
input_files = glob.glob('aligned_314.fasta')  # Adjust path as needed

for fasta_file in input_files:
    output_file = fasta_file.replace(".fasta", "_w_meta.fasta")
    
    with open(fasta_file, 'r') as input_handle, open(output_file, 'w') as output_handle:
        for record in SeqIO.parse(input_handle, 'fasta'):
            # Find the corresponding metadata (default to 'unknown' if not found)
            metadata = metadata_dict.get(record.id, {'date': 'unknown', 
                                        'wild_dom_lbm': 'unknown', 
                                        'state_domstat':'unknown',
                                        'states_grouped': 'unknown',
                                        'order': 'unknown'})
            # Update the header with all metadata fields
            record.id = f"{record.id}|{metadata['date']}|{metadata['wild_dom_lbm']}|{metadata['state_domstat']}|{metadata['states_grouped']}|{metadata['order']}|"
            record.description = ""  # Clear description to avoid duplicating info
            
            # Write the modified record to the output file
            SeqIO.write(record, output_handle, 'fasta')
    
    print(f"Modified FASTA file saved to {output_file}")
```
