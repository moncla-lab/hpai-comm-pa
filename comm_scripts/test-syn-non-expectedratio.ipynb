```python
from Bio import SeqIO
from Bio.Seq import Seq

##made with chatGPT

# Standard genetic code
genetic_code = {
    'TTT':'F','TTC':'F','TTA':'L','TTG':'L',
    'TCT':'S','TCC':'S','TCA':'S','TCG':'S',
    'TAT':'Y','TAC':'Y','TAA':'*','TAG':'*',
    'TGT':'C','TGC':'C','TGA':'*','TGG':'W',

    'CTT':'L','CTC':'L','CTA':'L','CTG':'L',
    'CCT':'P','CCC':'P','CCA':'P','CCG':'P',
    'CAT':'H','CAC':'H','CAA':'Q','CAG':'Q',
    'CGT':'R','CGC':'R','CGA':'R','CGG':'R',

    'ATT':'I','ATC':'I','ATA':'I','ATG':'M',
    'ACT':'T','ACC':'T','ACA':'T','ACG':'T',
    'AAT':'N','AAC':'N','AAA':'K','AAG':'K',
    'AGT':'S','AGC':'S','AGA':'R','AGG':'R',

    'GTT':'V','GTC':'V','GTA':'V','GTG':'V',
    'GCT':'A','GCC':'A','GCA':'A','GCG':'A',
    'GAT':'D','GAC':'D','GAA':'E','GAG':'E',
    'GGT':'G','GGC':'G','GGA':'G','GGG':'G'
}

bases = ['A', 'C', 'G', 'T']

def count_mutational_opportunities(seq):
    seq = seq.upper().replace("U", "T")

    synonymous = 0
    nonsynonymous = 0
    stop_gain = 0

    codons_processed = 0

    for i in range(0, len(seq) - 2, 3):
        codon = seq[i:i+3]

        if len(codon) != 3:
            continue
        if any(nt not in bases for nt in codon):
            continue
        if codon not in genetic_code:
            continue

        aa = genetic_code[codon]

        # Skip existing stop codons
        if aa == '*':
            continue

        codons_processed += 1

        for pos in range(3):
            for alt in bases:
                if alt == codon[pos]:
                    continue

                mutant = list(codon)
                mutant[pos] = alt
                mutant = ''.join(mutant)

                mutant_aa = genetic_code.get(mutant)

                if mutant_aa == '*':
                    stop_gain += 1
                elif mutant_aa == aa:
                    synonymous += 1
                else:
                    nonsynonymous += 1

    total = synonymous + nonsynonymous + stop_gain

    return {
        "codons": codons_processed,
        "synonymous": synonymous,
        "nonsynonymous": nonsynonymous,
        "stop_gain": stop_gain,
        "total_possible_single_nt_mutations": total
    }


# Example: read coding sequences from FASTA
fasta_file = "/Users/asjaeger/Downloads/ha_h5n1.fasta"

grand_total = {
    "codons": 0,
    "synonymous": 0,
    "nonsynonymous": 0,
    "stop_gain": 0,
    "total_possible_single_nt_mutations": 0
}

for record in SeqIO.parse(fasta_file, "fasta"):
    result = count_mutational_opportunities(str(record.seq))

    print(f"\n{record.id}")
    for k, v in result.items():
        print(f"{k}: {v}")

    for k in grand_total:
        grand_total[k] += result[k]

print("\n=== Genome Total ===")
for k, v in grand_total.items():
    print(f"{k}: {v}")
```
