```python
df25 = df[df['year_only'] == 2025]
df25.head()
```

```python
df25.drop(columns=['Unnamed: 0', 'genbank_acc', 'bp_length', 'ncbibvbrc_segment',
       'ncbibvbrc_nonambigdate', 'ncbibvbrc_statecoord', 'bvbrc_assembled','source',
       'Genome ID','strain_segment_updated', 'month_only', 'ncbibvbrc_hostgroup'], inplace=True)
df25
```

```python
df25.rename(columns={"strain": "strain_long", "ncbibvbrc_domesticstatus": "domestic_status"}, inplace=True)
```

```python
df25.columns = df25.columns.str.replace('ncbibvbrc_', '')
df25
```

```python
df25.division = df25.division.str.title()
df25.country = df25.country.str.title()
df25['flyway'] = df25['flyway'].str.replace('Fly', ' Fly', regex=False).str.strip()
df25['host'] = 'Avian'
df25['region'] = 'North America'

df25.subtype = df25.subtype.str.lower()


df25
```

```python
##i confirmed via nextclade that these are all 2344b, so will input there for h5_labelclade for all columns
df25['h5_label_clade'] = '2.3.4.4b'
df25
```

```python
##now i want to filter the fastas to only be the 2025 ones in this list
##and also rename the fasta headers to just strain not strain long:
from Bio import SeqIO
from pathlib import Path

# Create lookup: name -> strain
name_to_strain = dict(zip(df25["strain_long"], df25["strain"]))
valid_names = set(df25["strain_long"])

fasta_dir = Path("../phylo/2025_data_tomerge")        # input folder
out_dir = Path("../phylo/2025_data_tomerge/filtered_fastas") # output folder
out_dir.mkdir(exist_ok=True)

for fasta_file in fasta_dir.glob("*.fasta"):
    out_records = []

    for rec in SeqIO.parse(fasta_file, "fasta"):
        header = rec.id  # assumes header starts with name

        if header in valid_names:
            rec.id = name_to_strain[header]
            rec.description = ""  # avoid old header text
            out_records.append(rec)

    SeqIO.write(out_records, out_dir / fasta_file.name, "fasta")
```

```python
df25.to_csv('../phylo/2025_data_tomerge/filtered_fastas/cleaned_2025_meta.tsv', sep = '\t', index=False)
```
