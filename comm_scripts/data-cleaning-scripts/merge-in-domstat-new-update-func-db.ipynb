```python
import pandas as pd
from difflib import get_close_matches

# Step 1: Define canonical names (your known-good list)
canonical_states = [
    # US States
    'Alabama', 'Alaska', 'Arizona', 'Arkansas', 'California', 'Colorado',
    'Connecticut', 'Delaware', 'Florida', 'Georgia', 'Hawaii', 'Idaho',
    'Illinois', 'Indiana', 'Iowa', 'Kansas', 'Kentucky', 'Louisiana',
    'Maine', 'Maryland', 'Massachusetts', 'Michigan', 'Minnesota',
    'Mississippi', 'Missouri', 'Montana', 'Nebraska', 'Nevada',
    'New Hampshire', 'New Jersey', 'New Mexico', 'New York', 'North Carolina',
    'North Dakota', 'Ohio', 'Oklahoma', 'Oregon', 'Pennsylvania',
    'Rhode Island', 'South Carolina', 'South Dakota', 'Tennessee', 'Texas',
    'Utah', 'Vermont', 'Virginia', 'Washington', 'West Virginia',
    'Wisconsin', 'Wyoming',
    # Canadian Provinces
    'Alberta', 'British Columbia', 'Manitoba', 'New Brunswick',
    'Newfoundland', 'Nova Scotia', 'Nunavut', 'Ontario',
    'Prince Edward Island', 'Quebec', 'Saskatchewan',
    # Central America / Mexico
    'Costa Rica', 'Guatemala', 'Honduras', 'Panama',
    'Ciudad de Mexico', 'Mexico State', 'Durango', 'Sonora',
    # Other
    'Canada', 'Usa',
]

# Step 2: Manual overrides for cases fuzzy matching may get wrong
manual_map = {
    'Newyork':          'New York',
    'Southcarolina':    'South Carolina',
    'Northdakota':      'North Dakota',
    'Westvirginia':     'West Virginia',
    'Northcarolina':    'North Carolina',
    'Newjersey':        'New Jersey',
    'Newmexico':        'New Mexico',
    'Southdakota':      'South Dakota',
    'Newhampshire':     'New Hampshire',
    'Ciudaddemexico':   'Ciudad de Mexico',
    'Mexicostate':      'Mexico State',
}

# Step 3: Normalization function
def normalize_state(val):
    if pd.isna(val):
        return val
    val = str(val).strip()
    
    # Check manual overrides first (case-insensitive)
    title_val = val.title()
    if title_val in manual_map:
        return manual_map[title_val]
    
    # If already canonical, return as-is
    if val in canonical_states:
        return val
    
    # Try title-casing it first
    if title_val in canonical_states:
        return title_val

    # Fuzzy match as last resort
    matches = get_close_matches(title_val, canonical_states, n=1, cutoff=0.75)
    return matches[0] if matches else val  # return original if no match found

# Step 4: Apply
merged_df['state_clean'] = merged_df['division'].apply(normalize_state)

# Step 5: Audit anything that didn't resolve cleanly
unresolved = merged_df[merged_df['state_clean'] == merged_df['division']]['division'].unique()
print("Review these manually:", unresolved)
```

```python
merged_df['division'] = merged_df['state_clean']
merged_df.drop(columns=['state_clean'], inplace=True)
```
