```python
from Bio import SeqIO

def merge_fasta_files(file1, file2, output_file):
    sequences = []
    
    # Read the first FASTA file
    for record in SeqIO.parse('filtered_NA_ha.fasta', "fasta"):
        sequences.append(record)
    
    # Read the second FASTA file and concatenate the sequences
    for record in SeqIO.parse('filtered_ha.fasta', "fasta"):
        sequences.append(record)
    
    # Write the concatenated sequences to the output file
    SeqIO.write(sequences, 'filtered_merged.fasta', "fasta")
    print(f"Merged sequences written to {output_file}")

# Replace file paths with your input and output file paths
file1_path = '/file1'
file2_path = './file2'
output_path = './results/output_file'

merge_fasta_files(file1_path, file2_path, output_path)
```
